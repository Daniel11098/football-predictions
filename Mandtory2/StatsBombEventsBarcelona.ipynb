```python
from statsbombpy import sb
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sbn
```

```python
comps = sb.competitions()
```

```python
comps[comps["competition_name"] == "La Liga"].head(18)
```

```python
la_liga = comps[comps["competition_name"] == "La Liga"]
```

```python
la_liga.season_name
```

```python
print(len(la_liga))
```

```python
la_liga_splited= []
for _, c in la_liga.iterrows():
    matches= sb.matches(competition_id=c["competition_id"], season_id=c["season_id"])
    la_liga_splited.append(matches)

la_liga_df= pd.concat(la_liga_splited, ignore_index=True)
```

```python
la_liga_df.shape
```

```python
barcelonas_matches=la_liga_df[(la_liga_df["home_team"] == "Barcelona") | (la_liga_df["away_team"]== "Barcelona")]
```

```python
barcelonas_shots_splited= []
for _, event in barcelonas_matches.iterrows():
    barca_events= sb.events(match_id=event["match_id"])
    barca_shots= barca_events[barca_events["type"] == "Shot"]
    barcelonas_shots_splited.append(barca_shots)

barcelona_shots_df= pd.concat(barcelonas_shots_splited, ignore_index=True)
```
